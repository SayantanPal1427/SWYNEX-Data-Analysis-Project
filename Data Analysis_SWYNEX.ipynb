## DATA PREPARATION

```python
#Loading important packages
import numpy as np
import pandas as pd
from pandas.api.types import CategoricalDtype
```

```python
#load the data
data = pd.read_csv('c:/Users/hp/Desktop/CU_ML/Titanic-Dataset.csv')
data.head()
```

```python
#checking: is there any missing values?
data.isnull().sum()
# Since cabin has 687 missing values and it is also a less important factor so we redine our dataframe data after ommiting the cabin column.
# Age has 177 missing values but it is also an important factor so we can't omit that. We fill the missing value using the median age. 
# We omit those 2 rows for which we have those missing values in Embarked column.
```

```python
data = data.drop('Cabin',axis = 1)
data['Age'] = data['Age'].fillna(data['Age'].median())
data = data.dropna(subset=['Embarked'])
```

```python
data.isnull().sum()
# Now the dataframe is free from missing values 
```

```python
data['Survived'] = data['Survived'].astype('category') # cause it is a binary category
data['Sex'] = data['Sex'].astype('category') #cause it is a binary category
data['Embarked'] = data['Embarked'].astype('category') # it is a category with three factors S:southampton,england; C:cherbouge,France; Q:Queenstown,ireland

class_order = CategoricalDtype(categories = [3,2,1],ordered = True) # it is a ordinal category with 1:upper class , 2:middle class, 3:lower class
data['Pclass'] = data['Pclass'].astype(class_order)
```
